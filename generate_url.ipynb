```python
毕靖喜 23110941046
import random
import string
from IPython.display import HTML, display

# 1. 生成随机网址的函数
def generate_random_url(count=5):
    """
    生成指定数量的随机网址
    :param count: 要生成的网址数量，默认5个
    :return: 网址列表
    """
    chars = string.ascii_lowercase + string.digits
    domains = ["com", "cn", "net", "org", "io"]
    urls = []
    
    for _ in range(count):
        # 随机域名主体
        domain = ''.join(random.choice(chars) for _ in range(random.randint(5, 10)))
        suffix = random.choice(domains)
        # 随机路径
        path = ''.join(random.choice(chars) for _ in range(random.randint(3, 8)))
        url = f"https://www.{domain}.{suffix}/{path}"
        urls.append(url)
    return urls

# 2. 生成网址并保存到文件
urls = generate_random_url(10)
with open("urls.txt", "w", encoding="utf-8") as f:
    for url in urls:
        f.write(url + "\n")

print("✅ 网址已生成并保存到 urls.txt:")
for i, url in enumerate(urls, 1):
    print(f"{i}. {url}")

# 3. 在 Jupyter 里显示可点击的超链接
html_links = ""
for url in urls:
    html_links += f'<a href="{url}" target="_blank">{url}</a><br>'
display(HTML(html_links))
```
